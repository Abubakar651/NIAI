## 12. Normalization and Standardization

Min-Max: `(x - min) / (max - min)`

Z-score: `(x - mean) / std`

```python
x = np.array([10,20,30,40,50], dtype=float)
minmax=(x-np.min(x)/np.max(x)-np.min(x))
z=(x-np.mean(x))/np.std(x)
print("Minmax :",minmax)
print("z score : ",z)
print("new mean ",np.mean(z))
print("new std :",np.std(z))
```

## 16. ML Prediction: `y_hat = Xw + b`

```python
# One feature
X = np.array([1,2,3,4], dtype=float)
w = 10.0
b = 5.0
y_pred = X*w + b
print('Predictions:', y_pred)

# Multiple features
X = np.array([[1,2],[2,3],[3,4],[4,5]], dtype=float)
w = np.array([10,20], dtype=float)
b = 5.0
print('Multiple-feature predictions:', X @ w + b)
```

## 17. Loss Functions

MSE: `mean((actual - predicted)^2)`

MAE: `mean(abs(actual - predicted))`

```python
y_actual = np.array([10,20,30,40], dtype=float)
y_pred = np.array([12,18,33,37], dtype=float)
errors=y_actual-y_pred
squared_errors=errors**2
mse=np.mean(squared_errors)
mae=np.mean(np.abs(errors))
print('Errors:', errors)
print('Squared errors:', squared_errors)
print('MSE:', mse)
print('MAE:', mae)
```

## 18. Activation Functions

```python
x = np.array([-2,-1,0,1,2], dtype=float)

def sigmoid(x):
    return 1/(1+np.exp(-x))
def Relu(x):
    return np.maximum(0,x)
print ('Sigmoid :',sigmoid(x))
print('ReLu',Relu(x))
print("tanh ",np.tanh(x))
```

## 19. Softmax

Softmax converts class scores into probabilities whose sum is 1. Subtracting the maximum score improves numerical stability.

```python
def softmax(x):
    exp_x=np.exp(x-np.max(x))
    return exp_x/np.sum(exp_x)
scores = np.array([2.0,1.0,0.1])
probabilities = softmax(scores)
print('Probabilities:', probabilities)
print('Sum:', np.sum(probabilities))
```

## 23. Shuffle and Simple Train/Test Split

```python
rng=np.random.default_rng(42)
X=np.arange(20).reshape(10,2)
y=np.arange(10)

indices=rng.permutation(len(X))
train_size=int(0.8*10)
train_idx=indices[:train_size]
test_idx=indices[train_size:]
X_train,X_test=X[train_idx],X[test_idx]
y_train,y_test=y[train_idx],y[train_size]
print('X_train shape:', X_train.shape)  # Output: (8, 2)  -> 8 samples, 2 features
print('X_test shape:', X_test.shape)  
print('X_train :',X_train)  # Output: (2, 2)  -> 2 samples, 2 features
print('y_train:', y_train)              # 8 training labels
print('y_test:', y_test)                # 2 testing labels
```
