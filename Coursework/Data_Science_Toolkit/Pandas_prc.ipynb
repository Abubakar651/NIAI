```python
# Filter Age > 25
display(df[df['Age'] > 25])

# Filter Salary > 50,000
display(df[df['Salary'] > 50000])

# AND condition: both conditions must be true
display(df[(df['Age'] > 25) & (df['Salary'] > 50000)])

# OR condition: at least one condition must be true
display(df[(df['Age'] > 25) | (df['Salary'] > 50000)])

# NOT condition
display(df[~(df['Age']>25)])
```

```python
display(df[df["Department"].isin(["IT","HR"])])

display(df[~(df["Department"].isin(["IT","HR"]))])

display(df[df['Age'].between(25,30)])
display(df.query("Age > 25 and Salary > 50000"))
```

```python
df['AnnualSalary'] = df['Salary'] * 12

# Add a tax column
df['Tax'] = df['Salary'] * 0.10

df["Adult"]=np.where(df['Age']>=18,'Yes','No')
display(df)
```

```python
# Multiple conditions using np.select()
conditions=[
    df['Salary']>80000,
    df['Salary']>60000,
    df['Salary']>50000
]
choices=['High','Medium','Standard']
df['Salary_Catogart']=np.select(conditions,choices,default='low')
df
```

```python
# Create a small DataFrame containing missing values
missing_df=pd.DataFrame({
        'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha'],
    'Age': [25, np.nan, 28, 35],
    'Salary': [50000, 70000, np.nan, 90000]
})
display(missing_df.isna())

print(missing_df.isna().sum())
print(missing_df.isna().mean()*100)
display(missing_df.dropna())
display(missing_df.fillna(0))
ms=missing_df.isna().mean()*100
display(missing_df.fillna(ms))
```

```python
# Fill Age missing values with the median
missing_df['Age']=missing_df['Age'].fillna(
    missing_df['Age'].median()
)
# Fill Salary missing values with the mean
missing_df["Salary"]=missing_df['Salary'].fillna(
    missing_df['Salary'].mean()
)
display(missing_df)
```

```python
duplicate_df=pd.DataFrame({
    'Customer_id':[101, 102, 102, 103, 104, 104],
    'Name':['Ali', 'Sara', 'Sara', 'Ahmed', 'Ayesha', 'Ayesha']

})
print("Duplicated count",duplicate_df.duplicated().sum())
cleaned_dup=duplicate_df.drop_duplicates()
display(cleaned_dup)
display(duplicate_df.drop_duplicates(subset=['Customer_id']))
```

```python
text_df = pd.DataFrame({
    'Name': [' ali ', 'SARA', 'Ahmed Khan', 'Ayesha'],
    'Email': ['ALI@EXAMPLE.COM', 'sara@test.com', 'ahmed@test.com', 'ayesha@test.com']
})

# Convert to lowercase
text_df['Name_Lower'] = text_df['Name'].str.lower()

# Convert to uppercase
text_df['Name_Upper'] = text_df['Name'].str.upper()

# Capitalize
text_df['Name_Capitalized'] = text_df['Name'].str.capitalize()

# Remove leading/trailing spaces
text_df['Name_Clean'] = text_df['Name'].str.strip()

# String length
text_df['Name_Length'] = text_df['Name'].str.len()

display(text_df)
```

```python
#contains

display(
    text_df[
        text_df['Name'].str.contains('Ali', case=False, na=False)
    ]
)
display(
    text_df[
        text_df['Name'].str.strip().str.startswith('A',na=False)
    ]
)
display(
    
        text_df['Name'].str.replace('Ali','Ahmed',regex=False)
    
)
display(
    text_df['Name']
)
```
