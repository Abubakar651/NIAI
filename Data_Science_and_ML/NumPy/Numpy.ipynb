# NumPy for Machine Learning & Deep Learning

A practical, student-friendly NumPy notebook covering the most useful operations for ML/DL.

**Core idea:** NumPy provides fast numerical operations on vectors, matrices, and tensors.

```python
import numpy as np

# Check NumPy version
print('NumPy version:', np.__version__)
```

## 1. Creating Arrays

- 1D array → vector
- 2D array → matrix
- 3D+ array → tensor

```python
# 1D array / vector
x = np.array([10, 20, 30, 40, 50])
print(x)

# 2D array / matrix
X = np.array([[1, 2, 3], [4, 5, 6]])
print(X)

# 3D array / tensor
tensor = np.array([[[1,2],[3,4]], [[5,6],[7,8]]])
print(tensor)
```

## 2. Array Properties

These are essential for understanding ML/DL data dimensions.

```python
print('Shape:', X.shape)       # Rows and columns
print('Dimensions:', X.ndim)   # Number of axes
print('Size:', X.size)         # Total elements
print('Data type:', X.dtype)   # Element type
```

## 3. Creating Common Arrays

```python
print('Zeros:\n', np.zeros((3, 4)))      # All zeros
print('Ones:\n', np.ones((2, 3)))        # All ones
print('Identity:\n', np.eye(3))          # Identity matrix
print('Range:', np.arange(0, 10, 2))     # Start, stop, step
print('Linspace:', np.linspace(0, 1, 5)) # Evenly spaced values
```

## 4. Indexing and Slicing

```python
x = np.array([10, 20, 30, 40, 50])
print('First:', x[0])
print('Last:', x[-1])
print('First 3:', x[:3])
print('Every 2nd:', x[::2])

X = np.array([[10,20,30],[40,50,60],[70,80,90]])
print('First row:', X[0])
print('Second column:', X[:,1])
print('Bottom-right block:\n', X[1:,1:])
```

## 5. Boolean Filtering

```python
x = np.array([10, 25, 40, 55, 70])
print('Greater than 40:', x[x > 40])
print('Between 25 and 60:', x[(x >= 25) & (x <= 60)])
```

## 6. Vectorized Arithmetic

NumPy applies operations to whole arrays without explicit Python loops.

```python
x = np.array([1,2,3,4])
print('x + 10:', x + 10)
print('x * 2:', x * 2)
print('x squared:', x ** 2)
print('x / 2:', x / 2)

a = np.array([1,2,3])
b = np.array([10,20,30])
print('Element-wise addition:', a + b)
print('Element-wise multiplication:', a * b)
```

## 7. Broadcasting

Broadcasting lets NumPy combine arrays of compatible shapes. This is common when adding biases in ML/DL.

```python
X = np.array([[1,2,3],[4,5,6]])
print('Add scalar:\n', X + 10)

row_values = np.array([10,20,30])
print('Row-wise broadcasting:\n', X + row_values)
```

## 8. Mathematical Functions

```python
x = np.array([1,4,9,16], dtype=float)
print('sqrt:', np.sqrt(x))
print('exp:', np.exp(x))
print('log:', np.log(x))
print('abs:', np.abs(np.array([-5,2,-8])))
```

## 9. Statistics

```python
x = np.array([10,20,30,40,50])
print('Sum:', np.sum(x))
print('Mean:', np.mean(x))
print('Median:', np.median(x))
print('Min:', np.min(x))
print('Max:', np.max(x))
print('Std:', np.std(x))
print('Variance:', np.var(x))
print('25th percentile:', np.percentile(x,25))
print('75th percentile:', np.percentile(x,75))
```

## 10. Axis Operations

`axis=0` → calculate down rows / for each column.

`axis=1` → calculate across columns / for each row.

```python
X = np.array([[10,20,30],[40,50,60],[70,80,90]])
print('Column means:', np.mean(X, axis=0))
print('Row means:', np.mean(X, axis=1))
print('Column sums:', np.sum(X, axis=0))
print('Row sums:', np.sum(X, axis=1))
```

## 11. Reshape, Flatten and Transpose

```python
x = np.arange(12)
X = x.reshape(3,4)
print('Reshaped:\n', X)
print('Shape:', X.shape)
print('Flatten:', X.flatten())
print('Transpose:\n', X.T)
```

## 12. Normalization and Standardization

Min-Max: `(x - min) / (max - min)`

Z-score: `(x - mean) / std`

```python
x = np.array([10,20,30,40,50], dtype=float)
minmax = (x - np.min(x)) / (np.max(x) - np.min(x))
z = (x - np.mean(x)) / np.std(x)
print('Min-Max:', minmax)
print('Z-score:', z)
print('New mean:', np.mean(z))
print('New std:', np.std(z))
```

## 13. Linear Algebra: Vectors and Matrices

These operations are fundamental to linear regression, neural networks, embeddings, and deep learning layers.

```python
v = np.array([2,3,4])
print('Vector norm:', np.linalg.norm(v))

a = np.array([1,2,3])
b = np.array([4,5,6])
print('Dot product:', np.dot(a,b))
print('Dot product using @:', a @ b)

A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print('Element-wise A*B:\n', A*B)
print('Matrix multiplication A@B:\n', A@B)
```

## 14. Inverse, Pseudoinverse and Linear Equations

```python
A = np.array([[2,1],[1,2]], dtype=float)
print('Inverse:\n', np.linalg.inv(A))
print('Pseudoinverse:\n', np.linalg.pinv(A))

# Solve 2x + y = 5 and x + 3y = 6
b = np.array([5,6], dtype=float)
print('Solution:', np.linalg.solve(A,b))
```

## 15. Eigenvalues and Eigenvectors

Useful in advanced topics such as PCA.

```python
A = np.array([[2,1],[1,2]], dtype=float)
eigenvalues, eigenvectors = np.linalg.eig(A)
print('Eigenvalues:', eigenvalues)
print('Eigenvectors:\n', eigenvectors)
```

## 16. ML Prediction: `y_hat = Xw + b`

```python
# One feature
X = np.array([1,2,3,4], dtype=float)
w = 10.0
b = 5.0
y_pred = X*w + b
print('Predictions:', y_pred)

# Multiple features
X = np.array([[1,2],[2,3],[3,4],[4,5]], dtype=float)
w = np.array([10,20], dtype=float)
b = 5.0
print('Multiple-feature predictions:', X @ w + b)
```

## 17. Loss Functions

MSE: `mean((actual - predicted)^2)`

MAE: `mean(abs(actual - predicted))`

```python
y_actual = np.array([10,20,30,40], dtype=float)
y_pred = np.array([12,18,33,37], dtype=float)
errors = y_actual - y_pred
squared_errors = errors**2
mse = np.mean(squared_errors)
mae = np.mean(np.abs(errors))
print('Errors:', errors)
print('Squared errors:', squared_errors)
print('MSE:', mse)
print('MAE:', mae)
```

## 18. Activation Functions

```python
x = np.array([-2,-1,0,1,2], dtype=float)

def sigmoid(x):
    # Maps values approximately into 0 to 1
    return 1 / (1 + np.exp(-x))

def relu(x):
    # ReLU = max(0, x)
    return np.maximum(0, x)

print('Sigmoid:', sigmoid(x))
print('ReLU:', relu(x))
print('Tanh:', np.tanh(x))
```

## 19. Softmax

Softmax converts class scores into probabilities whose sum is 1. Subtracting the maximum score improves numerical stability.

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

scores = np.array([2.0,1.0,0.1])
probabilities = softmax(scores)
print('Probabilities:', probabilities)
print('Sum:', np.sum(probabilities))
```

## 20. One-Hot Encoding and Classification

```python
labels = np.array([0,2,1,2])
num_classes = 3
one_hot = np.eye(num_classes)[labels]
print('One-hot:\n', one_hot)

probabilities = np.array([[0.1,0.7,0.2],[0.8,0.1,0.1],[0.2,0.3,0.5]])
print('Predicted classes:', np.argmax(probabilities, axis=1))
```

## 21. Random Numbers and Reproducibility

```python
rng = np.random.default_rng(42)
print('Random floats:', rng.random(5))
print('Random integers:', rng.integers(0,10,size=5))

samples = rng.normal(loc=0, scale=1, size=1000)
print('Normal sample mean:', np.mean(samples))
print('Normal sample std:', np.std(samples))
```

## 23. Shuffle and Simple Train/Test Split

```python
rng = np.random.default_rng(42)
X = np.arange(20).reshape(10,2)
y = np.arange(10)
indices = rng.permutation(len(X))
train_size = int(0.8*len(X))
train_idx = indices[:train_size]
test_idx = indices[train_size:]
X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
print('X_train shape:', X_train.shape)
print('X_test shape:', X_test.shape)
print('y_train:', y_train)
print('y_test:', y_test)
```

## 24. Conditional Logic: `where` and `select`

```python
salary = np.array([40000,55000,65000,90000])
category = np.where(salary >= 60000, 'High', 'Standard')
print('np.where:', category)

conditions = [salary >= 80000, salary >= 60000, salary >= 50000]
choices = ['High','Medium','Standard']
print('np.select:', np.select(conditions, choices, default='Low'))
```

## 25. Missing Values and Cleaning

```python
x = np.array([10,20,np.nan,40,50])
print('Is NaN:', np.isnan(x))
print('NaN count:', np.sum(np.isnan(x)))
print('Mean ignoring NaN:', np.nanmean(x))

# Replace NaN with the mean
x_clean = np.nan_to_num(x, nan=np.nanmean(x))
print('Cleaned:', x_clean)
```

## 26. Unique Values and Counts

```python
labels = np.array(['cat','dog','cat','bird','dog','cat'])
values, counts = np.unique(labels, return_counts=True)
print('Values:', values)
print('Counts:', counts)
```

## 27. Clipping, Sorting and Indexing

```python
x = np.array([10,20,30,100,200])
print('Clipped:', np.clip(x, 0, 50))

x = np.array([40,10,30,20])
print('Sorted:', np.sort(x))
indices = np.argsort(x)
print('Sort indices:', indices)
print('Sorted by indices:', x[indices])
```

## 28. Concatenation and Stacking

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print('Concatenate:', np.concatenate([a,b]))

A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print('Vertical stack:\n', np.vstack([A,B]))
print('Horizontal stack:\n', np.hstack([A,B]))
```

## 29. Cumulative Operations

```python
sales = np.array([100,150,120,200])
print('Cumulative sum:', np.cumsum(sales))
print('Cumulative product:', np.cumprod(np.array([1,2,3,4])))
```

## 30. Numerical Derivative Approximation

`np.gradient()` gives a finite-difference approximation to a derivative. This helps connect calculus to ML optimization.

```python
x = np.linspace(-3,3,7)
y = x**2
print('x:', x)
print('f(x):', y)
print('Approx derivative:', np.gradient(y,x))
```

## 31. Correlation and Covariance

```python
x = np.array([1,2,3,4,5])
y = np.array([2,4,5,8,10])
print('Correlation matrix:\n', np.corrcoef(x,y))
print('Covariance matrix:\n', np.cov(x,y))
```

## 32. Simple Neural-Network Layer

A basic dense layer is:

`Z = X @ W + b`

followed by an activation such as ReLU.

```python
X = np.array([[1.,2.],[2.,3.],[3.,4.]])
W = np.array([[0.5,0.2,0.1],[0.4,0.3,0.2]])
b = np.array([0.1,0.2,0.3])

Z = X @ W + b
A = np.maximum(0, Z)
print('Z:\n', Z)
print('ReLU output:\n', A)
```

## ML/DL mental model

`Raw Data → NumPy Arrays → Shape/Reshape → Cleaning → Scaling → Vectors/Matrices → X @ W + b → Activation → Prediction → Loss → Gradient → Parameter Update`
